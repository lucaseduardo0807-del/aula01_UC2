```python
#%pip install pandas
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'C:\Users\lucas.emoura\Documents\PastaLucas\aula01_UC2\aula02\01.amazon_sales_dataset.csv')
df.columns
```

```python
COLUNA = 'total_sales'
```

```python
coluna_array = np.array(df[COLUNA])

media = np.mean(coluna_array)
mediana = np.median(coluna_array)
max = np.max(coluna_array)
min = np.min(coluna_array)
amplitude = max - min
delta_media_mediana = media-mediana
div_media_mediana = media/mediana
q1 = np.percentile(coluna_array, 0.25)
q3 = np.percentile(coluna_array, 0.75)
iqr = q3 - q1

desvio = np.std(coluna_array)
cv = (desvio/media)*100
```

```python
print(f'Media: {media:.2f}')
print(f'Mediana: {mediana:.2f}')
print(f'Delta Media-Mediana: {delta_media_mediana:.2f}')
print(f'Razão Media-Mediana: {div_media_mediana:.2f}')
```

```python
# medidas de dispersão/posição
print(f'Amplitude: {amplitude:.2f}')
print(f'Desvio Padrão: {desvio:.2f}')
print(f'Coeficiente de Variação (CV): {cv:.2f}')
print(30*'-')
print(f'Mínimo: {min:.2f}')
print(f'Q1: {q1:.2f}')
print(f'Q2: {mediana:.2f}')
print(f"Q3: {q3:.2f}")
print(f'Máximo: {max:2f}')
print(f'IQR: {iqr:.2f}')
```
