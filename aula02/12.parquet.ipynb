```python
hora_inicio = dt.datetime.now()

# falar a pasta
lista_arquivos = os.listdir(DADOS)
print(lista_arquivos)
for arquivo in lista_arquivos:
    print(f'Lendo arquivo: {arquivo}')
    df = pl.read_csv(DADOS+arquivo, separator=';',
                     encoding='iso-8859-1')
    # se existe df_bf nas variaveis locais
    if 'df_bf' in locals():
        # juntar os meses (datasets/dataframes)
        df_bf = pl.concat([df_bf, df])
    else:
        df_bf = df
        # primeira passada do for, não tem df_bf, então eu
        # crio e escrevo o df nele
        
    # deletar o mes atual, pq ja escrevi ele na variavel 
    # que vou salvar
    del df

df_bf = df_bf.with_columns(
    pl.col('VALOR PARCELA').str.replace(',', '.')
    .cast(pl.Float64)
)

df_bf.write_parquet(r'C:\Users\lucas.emoura\Documents\PastaLucas\aula01_UC2\aula02\dados_bronze\df_bf.parquet')
```
