```python
# %pip install pandas
import pandas as pd
import numpy as np
```

```python
ARQUIVO = r'C:\Users\lucas.emoura\Documents\PastaLucas\aula01_UC2\aula02\03.BaseDPEvolucaoMensalCisp.csv'
df = pd.read_csv(ARQUIVO,
                 sep=';',
                 encoding='IBM860')
df.head(10)
```

```python
# agregar por município -> analisar roubo de veículo

df_roubo_munic = df [[
    'munic', 'roubo_veiculo'
]]

# agrupar
df_roubo_munic_agg = (
    df_roubo_munic.groupby('munic')
    .sum()
    .reset_index()
    .sort_values(by='roubo_veiculo', ascending=False)
)
df_roubo_munic_agg
```

```python
roubo_array = np.array(df_roubo_munic_agg['roubo_veiculo'])

media = np.mean(roubo_array)
mediana = np.mean(roubo_array)
diff_media_mediana = media - mediana
div_media_mediana = media/mediana

desvio = np.std(roubo_array)
cv = desvio/media

q1 = np.percentile(roubo_array, 25)
q3 = np.percentile(roubo_array, 75)
iqr = q3 - q1

max = np.max(roubo_array)
min = np.min(roubo_array)
amplitude = max - min
```

```python
assimetria = df['roubo_veiculo'].skew()
curtose = df['roubo_veiculo'].kurtosis()

print(f'Assimetria: {assimetria:.4f}')
print(f'Curtose: {curtose:.4f}')
```
