```python
# 1. QUANTIDADE DE VEÍCULOS RECUPERADOS POR MUNICÍPIO
```

```python
df_munic = df_recuperacao_veiculos[["munic","recuperacao_veiculos"]]

df_munic_agrupado = (
    df_munic.groupby('munic')['recuperacao_veiculos']
    .sum().reset_index().sort_values(by="recuperacao_veiculos", ascending=False)
)

df_munic_agrupado.head(10)
```

```python
# 2. ANÁLISE POR ANO
```

```python
df_ano = df_recuperacao_veiculos[["ano","roubo_veiculo"]]

df_roubo_agrupado = (
    df_ano.groupby('ano')['roubo_veiculo']
    .sum().reset_index().sort_values(by="ano", ascending=False)
)

df_roubo_agrupado.head(50)
```

```python
df_ano = df_recuperacao_veiculos[["ano","recuperacao_veiculos"]]

df_ano_agrupado = (
    df_ano.groupby('ano')['recuperacao_veiculos']
    .sum().reset_index().sort_values(by="ano", ascending=False)
)

df_ano_agrupado.head(50)
```

```python
plt.subplots(1, 2, figsize=(16, 7))

plt.suptitle('Roubo e Recuperação de Veículos', fontsize=16)

plt.subplot(1, 2, 1)
plt.bar(df_roubo_agrupado['ano'],
         df_roubo_agrupado['roubo_veiculo'],
         color='black')
plt.title('Numero de Veículos roubados')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
plt.bar(df_ano_agrupado['ano'],
        df_ano_agrupado['recuperacao_veiculos'],
        color='blue')
plt.title('Número de Veículos Recuperados')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
